# 프롬프트 구성 > LCEL 파이프라인 구성 > invoke

금융도메인 프롬프트 : system(역할, 페르소나), user, ai(퓨샷러닝), user 로 구성합니다.

```python
# ── 프롬프트 구성: system / user / ai / user ──
chat_template = ChatPromptTemplate.from_messages([
    (
        "system",
        """당신은 CFA(국제공인재무분석사) 자격을 보유한 퀀트 애널리스트입니다.
다음 원칙을 반드시 따릅니다:
1. 데이터 기반 분석 — 감이 아닌 수치로 판단합니다.
2. 리스크 우선 사고 — 수익보다 손실 가능성을 먼저 언급합니다.
3. 멀티 프레임 분석 — 거시(매크로) → 섹터 → 종목 순으로 분석합니다.
4. 시나리오 3단계 — 낙관 / 기본 / 비관 시나리오를 항상 제시합니다.
5. 답변 형식 — 마크다운 표와 불릿포인트를 활용해 구조화합니다."""
    ),
    (
        "human",
        """{asset_type} 종목 {ticker}에 대해 분석해줘.
현재 보유 금액은 {investment_amount}원이고,
투자 목적은 {investment_goal}이야.
리스크 허용 범위는 {risk_level}이야."""
    ),
    (
        "ai",
        """알겠습니다. {ticker} 분석을 시작하겠습니다.
거시경제 환경 → 섹터 포지션 → 종목 펀더멘털 순으로 검토한 후,
{risk_level} 리스크 프로파일에 맞는 3가지 시나리오를 제시하겠습니다.
{investment_goal} 목표 달성을 위한 포지션 전략도 함께 드리겠습니다."""
    ),
    (
        "human",
        """{additional_question}
추가로, 지금 시장 상황에서 {asset_type} 포트폴리오 내 {ticker} 비중을
몇 %로 가져가는 게 최적인지 수치 근거와 함께 알려줘."""
    ),
])
```

# ── LCEL 파이프라인 ──

```python
# ── LCEL 파이프라인 ──
parser = StrOutputParser()

chain = chat_template | llm | parser
```

# ── invoke ──

```python
# ── invoke ──
response = chain.invoke({
    "asset_type"        : "국내 성장주",
    "ticker"            : "삼성전자 (005930)",
    "investment_amount" : "5,000만",
    "investment_goal"   : "3년 내 자산 2배 달성",
    "risk_level"        : "중위험 (MDD 최대 -20% 허용)",
    "additional_question": """
현재 반도체 사이클 위치와 HBM 수혜 가능성을 근거로
향후 6개월 주가 밴드를 제시해줘.
또한 환율(원/달러), 미국 금리, 중국 수요 변수별
민감도 분석표도 만들어줘."""
})

print(response)
```
