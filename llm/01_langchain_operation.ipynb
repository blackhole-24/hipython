# invoke [단일 입력 단일 응답]

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")

# invoke() - 단일 입력, 단일 응답 (동기)
response = llm.invoke("삼성전자 주가에 영향을 주는 요인은?")
print(response.content)
```

# batch() - 여러 입력을 한 번에 처리 

```python
# batch() - 여러 입력을 한 번에 처리 (동기)
# 중복 토큰을 줄여 묶어 처리 → 비용 절감 효과
questions = [
    "삼성전자 주가에 영향을 주는 요인은?",
    "코스피 지수란 무엇인가?",
    "ETF와 펀드의 차이는?"
]
responses = llm.batch(questions)
for r in responses:
    print(r.content)
    print()
```

# stream() - 실시간 스트리밍 출력 

```python
# stream() - 실시간 스트리밍 출력 (토큰 단위로 출력)
for chunk in llm.stream("환율이 수출 기업에 미치는 영향은?"):
    print(chunk.content, end="", flush=True)
```

# 동기 vs. 비동기

```python
import asyncio
import time
from langchain_openai import ChatOpenAI
#from langchain_core.messages import HumanMessage

#llm = ChatOpenAI(model="gpt-4o-mini")

questions = [f"주식 투자 팁 {i}번째" for i in range(10)]

# 비동기 병렬 처리
async def invoke_async(question): # 각 프롬프트 1개 처리
    response = await llm.ainvoke(question)
    return response.content

async def invoke_parallel(questions): # 다수의 프롬프트 목록을 처리
    tasks = [invoke_async(q) for q in questions]
    return await asyncio.gather(*tasks)
```

```python
res = await invoke_async(questions[0])
#print(res)   
```

# 비동기 실행 시간 측정

```python
# 비동기 실행 시간 측정
start = time.perf_counter()
results = await invoke_parallel(questions)
end = time.perf_counter()
print(f"비동기 처리: {end - start:.2f}초")   # 약 5~6초
```

```python
# 동기 실행 시간 측정
start = time.perf_counter()
for q in questions:
    response = llm.invoke(q) #동기호출
end = time.perf_counter()
print(f"동기 처리: {end - start:.2f}초")     # 약 40~50초
```

# 토큰 사용량 확인 

```python
from langchain_community.callbacks import get_openai_callback

with get_openai_callback() as callback:
    message = [HumanMessage(content="서울 광장시장에서 가장 맛있는 길거리 음식은?")]
    response = llm.invoke(message)
    print(response.content)
    print(f"\n총 토큰: {callback.total_tokens}")
    print(f"프롬프트 토큰: {callback.prompt_tokens}")
    print(f"응답 토큰: {callback.completion_tokens}")
    print(f"비용(USD): ${callback.total_cost:.6f}")
```

# 대화시작 / 후속질문 [삼성전자]

```python
# 대화시작
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="당신은 주식 투자 전문가로 고객의 투자 결정에 도움을 줍니다."),
    HumanMessage(content="삼성전자 주식을 지금 매수해도 될까요?")
]

response = llm.invoke(messages)
print(response.content)
```

```python
# 후속질문
messages = messages + [response, HumanMessage(content="그렇다면 목표 주가는 얼마로 보시나요?")]

response2 = llm.invoke(messages)
print(response2.content)
```
