```python
import pandas as pd

# 샘플 데이터프레임 생성
data = {
    '이름': ['김철수', '이영희', '박민수', '정수진', '최영호', '한미래', '윤지원'],
    '나이': [25, 28, 22, 31, 27, 24, 29],
    '점수': [85, 92, 78, 95, 88, 73, 91]
}
df = pd.DataFrame(data)

# 처음 5개 행 보기
print("=== head() ===")
print(df.head())

# 마지막 3개 행 보기
print("\n=== tail(3) ===")
print(df.tail(3))

# 데이터프레임 기본 정보
print("\n=== info() ===")
print(df.info())

# 기술 통계량
print("\n=== describe() ===")
print(df.describe())
```

```python
#인덱싱

 [행, 열] 인덱싱

 * iloc[] 행기반 인덱싱 >>> [행, 열]
 * loc[] 이름기반(위치) 인덱싱 >>> [행이름, 열이름]

 - : >> start:end 까지 슬라이싱
 - loc[행이름, 열이름]
 - iloc[행번호, 열번호]
```

```python
#df1[['이름,'나이']]
df1.loc[:,['이름','나이']] # loc[행, 열] 행과 열을 반드시 지정
```

```python
df1.loc['A']
```

```python
df1.iloc[0:3, 0:2] #행번호, 열
```

```python
# 연습문제
```

```python
# [기본]

# 모든 행에서 이름 열만 선택하세요.
df1['이름']

# 모든 행에서 이름과 score 열을 선택하세요.
df1[['이름', 'score']]

# 인덱스 C 행의 모든 데이터를 선택하세요.
df1.loc['C']

# 인덱스 D 행의 age 값만 선택하세요.
df1.loc['D', 'age']
```

```python
# [슬라이싱]

# 인덱스 B부터 E까지의 모든 행을 선택하세요.
df1.loc['B':'E']
# 인덱스 C부터 F까지의 이름과 age 열만 선택하세요.
df1.loc['C':'F', ['이름','age']]
# 인덱스 A부터 G까지에서 score 열만 선택하세요.
df1.loc['A':'G', 'score']
```

Boolean 인덱싱

```python
df [ df['점수'] > 90 ] #vectorization boolean list의 요소 값 True, False
```

```python
cond = df1['나이'] >= 27
```

```python
# [조건 + loc]

# age가 27 이상인 행만 선택하세요.
df1.loc[cond]

# score가 90 이상인 사람의 이름과 score만 출력하세요.
df1.loc[  df1['점수']>=90,  ['이름', '점수']  ]

# age가 25 이상이면서 score가 90 이상인 행을 모두 선택하세요.
cond1 = df1['나이'] >=25
cond2 = df1['점수']>=90
df1.loc[cond1 & cond2]
```

```python
iloc 인덱싱 & 슬라이싱 연습문제

👉 정수 위치 기반 (0부터 시작)
```

```python
# [기본]
# 첫 번째 행의 모든 데이터를 선택하세요.
df1.iloc[0]
# 세 번째 행의 이름 값만 선택하세요.
df1.iloc[2,0]
# 처음 3개의 행을 선택하세요.
df1.iloc[ :3, ]
# 마지막 행의 score 값만 선택하세요.
df1.iloc[-1, 2]
```

```python
# [슬라이싱]
# 1번 행부터 4번 행까지 선택하세요.
df1.iloc[1:5]
# 2번 행부터 5번 행까지에서 이름과 age 열만 선택하세요.
df1.iloc[2:6, [0,1]]
# 모든 행에서 age 열만 선택하세요.
df1.iloc[:, 1]  
```

```python
# [조합 문제]
# score 열만 선택하되, 앞에서 4개의 행만 출력하세요.
df1.iloc[0:4,2]
# 이름과 score 열을 선택하되, 마지막 3개의 행만 출력하세요.
df.iloc[-3: , ::2] #start:end:step
df1.tail(3).loc[:,['이름','점수']]
# 짝수 번째 행(0, 2, 4, 6)의 모든 데이터를 선택하세요.
df1.iloc[[0,2,4,6]]
```

```python
# 행주가와 열추가
```

```python
df.loc['H'] = ['장민호', 26, 88]
df['grade'] = ['A','B','A', 'A','B','C','A']
```

```python
# 인덱스로 행 삭제
df.drop('A', axis=0)  # 'A' 인덱스의 행 삭제

# 여러 행 한 번에 삭제
df.drop(['B', 'C'], axis=0)
```

```python
# 열 이름으로 열 삭제
df.drop('grade', axis=1)  # 'grade' 열 삭제

# 여러 열 한 번에 삭제
df.drop(['age', 'score'], axis=1)
```

```python
# 데이터형 변환
```

```python
# 객체를 숫자형으로 변환
df = df.astype({'age': int, 'score': float})

# 문자열로 변환
df['student_id'] = df['student_id'].astype(str)

# 날짜형으로 변환
df['date'] = pd.to_datetime(df['date'])
```

```python
# 인덱스의 설정
```

```python
# 'student_id' 열을 인덱스로 설정
df.set_index('student_id', inplace=True)

# 인덱스를 다시 열로 되돌리기
df.reset_index(inplace=True)
```
