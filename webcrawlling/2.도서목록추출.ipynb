```python
import requests
from bs4 import BeautifulSoup

# 셀렉터 정의
selector       = "#yesBestList div.item_info"
title_selector = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"   # ✅ 누락된 price_selector 추가

# Book 클래스 정의
class Book:
    def __init__(self, title, author, price):
        self.title  = title
        self.author = author
        self.price  = price

    def __str__(self):
        return f"{self.title} / {self.author} / {self.price}"

# 페이지 지정 (1~5페이지)
base_url = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&page='

for page in range(1, 6):
    url = base_url + str(page)
    soup = BeautifulSoup(requests.get(url).text, 'html.parser')
    items = soup.select(selector)

    for i, item in enumerate(items):
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)

        title_val  = title.text.strip()  if title  else 'N/A'
        author_val = author.text.strip() if author else 'N/A'
        price_val  = price.text.strip()  if price  else 'N/A'

        book = Book(title_val, author_val, price_val)
        print(f"{(page-1)*20 + i+1}. {book}")
```

2. 한 페이지의 도서 목록 추출 > 리스트

```python
import requests
from bs4 import BeautifulSoup

url = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001'
soup = BeautifulSoup(requests.get(url).text, 'html.parser')

selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

items = soup.select(selector)

# 도서 목록을 딕셔너리 리스트로 저장
book_list = []
for item in items:
    title  = item.select_one(title_selector)
    author = item.select_one(author_selector)
    price  = item.select_one(price_selector)

    book_list.append({
        "제목" : title.text.strip()  if title  else "N/A",
        "저자" : author.text.strip() if author else "N/A",
        "가격" : price.text.strip()  if price  else "N/A"
    })

# 결과 출력
for i, book in enumerate(book_list, start=1):
    print(f"{i}. {book['제목']} / {book['저자']} / {book['가격']}")

print(f"\n총 {len(book_list)}권 추출")
```

3. 여러 페이지의 도서 목록 추출

```python
import requests
from bs4 import BeautifulSoup

base_url        = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&page='
selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

all_books = []  # 전체 페이지 도서 누적 리스트

for page in range(1, 6):  # 1~5페이지
    soup = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select(selector)

    for item in items:
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)

        all_books.append({
            "제목" : title.text.strip()  if title  else "N/A",
            "저자" : author.text.strip() if author else "N/A",
            "가격" : price.text.strip()  if price  else "N/A"
        })

    print(f"✅ {page}페이지 완료 ({len(items)}권)")

# 결과 출력
print(f"\n전체 {len(all_books)}권 추출")
for i, book in enumerate(all_books, start=1):
    print(f"{i}. {book['제목']} / {book['저자']} / {book['가격']}")
```

4. Book 클래스를 이용한 객체 생성

```python
import requests
from bs4 import BeautifulSoup

# Book 클래스 정의
class Book:
    def __init__(self, title, author, price):
        self.title  = title
        self.author = author
        self.price  = price

    def __str__(self):
        return f"제목: {self.title} | 저자: {self.author} | 가격: {self.price}"

base_url        = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&page='
selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

book_objects = []  # Book 객체 리스트

for page in range(1, 6):
    soup = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select(selector)

    for item in items:
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)

        book = Book(
            title.text.strip()  if title  else "N/A",
            author.text.strip() if author else "N/A",
            price.text.strip()  if price  else "N/A"
        )
        book_objects.append(book)

# 결과 출력
print(f"총 {len(book_objects)}개 Book 객체 생성\n")
for i, book in enumerate(book_objects, start=1):
    print(f"{i}. {book}")

# 특정 책 접근 예시
print(f"\n첫 번째 책 제목: {book_objects[0].title}")
print(f"첫 번째 책 저자: {book_objects[0].author}")
```
