2. 한 페이지의 도서 목록 추출 > 리스트

```python
import requests
from bs4 import BeautifulSoup

url = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001'
soup = BeautifulSoup(requests.get(url).text, 'html.parser')

selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

items = soup.select(selector)

# 도서 목록을 딕셔너리 리스트로 저장
book_list = []
for item in items:
    title  = item.select_one(title_selector)
    author = item.select_one(author_selector)
    price  = item.select_one(price_selector)

    book_list.append({
        "제목" : title.text.strip()  if title  else "N/A",
        "저자" : author.text.strip() if author else "N/A",
        "가격" : price.text.strip()  if price  else "N/A"
    })

# 결과 출력
for i, book in enumerate(book_list, start=1):
    print(f"{i}. {book['제목']} / {book['저자']} / {book['가격']}")

print(f"\n총 {len(book_list)}권 추출")
```

3. 여러 페이지의 도서 목록 추출

```python
import requests
from bs4 import BeautifulSoup

base_url        = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&page='
selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

all_books = []  # 전체 페이지 도서 누적 리스트

for page in range(1, 6):  # 1~5페이지
    soup = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select(selector)

    for item in items:
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)

        all_books.append({
            "제목" : title.text.strip()  if title  else "N/A",
            "저자" : author.text.strip() if author else "N/A",
            "가격" : price.text.strip()  if price  else "N/A"
        })

    print(f"✅ {page}페이지 완료 ({len(items)}권)")

# 결과 출력
print(f"\n전체 {len(all_books)}권 추출")
for i, book in enumerate(all_books, start=1):
    print(f"{i}. {book['제목']} / {book['저자']} / {book['가격']}")
```

4. Book 클래스를 이용한 객체 생성

```python
import requests
from bs4 import BeautifulSoup

# Book 클래스 정의
class Book:
    def __init__(self, title, author, price):
        self.title  = title
        self.author = author
        self.price  = price

    def __str__(self):
        return f"제목: {self.title} | 저자: {self.author} | 가격: {self.price}"

base_url        = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&pageNumber='
selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

book_objects = []  # Book 객체 리스트

for page in range(1, 7):
    soup = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select(selector)

    for item in items:
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)

        book = Book(
            title.text.strip()  if title  else "N/A",
            author.text.strip() if author else "N/A",
            price.text.strip()  if price  else "N/A"
        )
        book_objects.append(book)

# 결과 출력
print(f"총 {len(book_objects)}개 Book 객체 생성\n")
for i, book in enumerate(book_objects, start=1):
    print(f"{i}. {book}")

# 특정 책 접근 예시
print(f"\n첫 번째 책 제목: {book_objects[0].title}")
print(f"첫 번째 책 저자: {book_objects[0].author}")
```

```python
import requests
from bs4 import BeautifulSoup

base_url = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&page='

all_books = []

for page in range(1, 6):  # 1~5페이지
    soup  = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select("#yesBestList li")  # ✅ div.item_info → li 로 변경 (중복 방지)

    for item in items:
        # 중복 li 필터링 (data-goods-no 없는 항목 제외)
        if not item.get("data-goods-no"):
            continue

        title  = item.select_one("div.info_name a.gd_name")
        author = item.select_one("span.info_auth a")
        pub    = item.select_one("span.info_pub a")
        date   = item.select_one("span.info_date")

        # ✅ 가격 셀렉터 수정
        price  = item.select_one("em.yes_b") or \
                 item.select_one("strong.txt_num em") or \
                 item.select_one("span.info_price em")

        all_books.append({
            "제목"  : title.text.strip()  if title  else "N/A",
            "저자"  : author.text.strip() if author else "N/A",
            "출판사": pub.text.strip()    if pub    else "N/A",
            "출판일": date.text.strip()   if date   else "N/A",
            "가격"  : price.text.strip()  if price  else "N/A"
        })

    print(f"✅ {page}페이지 완료 ({len(items)}권)")

# 결과 출력
print(f"\n전체 {len(all_books)}권 추출\n")
for i, book in enumerate(all_books, start=1):
    print(f"[{i}위] {book['제목']} / {book['저자']} / {book['출판사']} / {book['출판일']} / {book['가격']}")
```

```python
import requests
import sqlite3
from bs4 import BeautifulSoup

# Book 클래스 정의
class Book:
    def __init__(self, title, author, price):
        self.title  = title
        self.author = author
        self.price  = price

# books 리스트 생성
base_url        = 'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&pageNumber='
selector        = "#yesBestList div.item_info"
title_selector  = "div.info_name a.gd_name"
author_selector = "span.info_auth a"
price_selector  = "span.info_price strong"

books = []

for page in range(1, 7):
    soup  = BeautifulSoup(requests.get(base_url + str(page)).text, 'html.parser')
    items = soup.select(selector)
    for item in items:
        title  = item.select_one(title_selector)
        author = item.select_one(author_selector)
        price  = item.select_one(price_selector)
        books.append(Book(
            title.text.strip()  if title  else "N/A",
            author.text.strip() if author else "N/A",
            price.text.strip()  if price  else "0"
        ))

# DB 연결 및 테이블 생성
conn   = sqlite3.connect('_books.db')
cursor = conn.cursor()
cursor.execute('''
CREATE TABLE IF NOT EXISTS books (
    title  TEXT,
    author TEXT,
    price  TEXT
)
''')
```

```python
# 데이터 삽입
for book in books:
    cursor.execute('''
    INSERT INTO books (title, author, price) VALUES (?, ?, ?)
    ''', (book.title, book.author, book.price))

conn.commit()
cursor.close()
conn.close()
```
