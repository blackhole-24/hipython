# 4 다국어 감성 분석 모델

```python
# 4번 다국어 감성 분석 모델
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="tabularisai/multilingual-sentiment-analysis")

texts = [
    "Trump warns of extremely powerful military strike against Iran within weeks",  # 영어 (부정)
    "트럼프, 수주 내 이란에 대한 극도로 강력한 군사 타격 경고",                             # 한국어 (부정)
    "トランプ氏、数週間以内にイランへの強力な軍事攻撃を警告",                                 # 일본어 (부정)
    "BTS announces surprise world tour with free tickets for fans",                 # 영어 (긍정)
    "방탄소년단, 팬들에게 무료 티켓 제공하며 깜짝 월드투어 발표",                             # 한국어 (긍정)
    "BTS、ファンに無料チケットを提供しサプライズワールドツアーを発表",                           # 일본어 (긍정)
    "Hyundai Motor recalls 500,000 vehicles due to engine defect",                  # 영어 (부정)
    "현대자동차, 엔진 결함으로 50만대 리콜 결정",                                        # 한국어 (부정)
    "現代自動車、エンジン欠陥で50万台をリコール決定",                                      # 일본어 (부정)
]

# Expected: 부정, 부정, 부정, 긍정, 긍정, 긍정, 부정, 부정, 부정

results = classifier(texts)
for text, result in zip(texts, results):
    print(f"입력: {text}")
    print(f"결과: {result['label']} ({result['score']:.4f})")
    print()
```

# 5 다국어 BERT

```python
# 5번 빈칸 채우기 (다국어 BERT)
from transformers import pipeline

unmasker = pipeline("fill-mask", model="bert-base-multilingual-cased")

# 영어: Tesla is the world's largest [MASK] vehicle manufacturer.
result_en = unmasker("Tesla is the world's largest [MASK] vehicle manufacturer.")
print("=== 영어 ===")
print("입력: Tesla is the world's largest [MASK] vehicle manufacturer.")
for r in result_en[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 한국어: 삼성전자는 세계 최대의 [MASK] 기업이다.
result_kr = unmasker("삼성전자는 세계 최대의 [MASK] 기업이다.")
print("=== 한국어 ===")
print("입력: 삼성전자는 세계 최대의 [MASK] 기업이다.")
for r in result_kr[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 일본어: 東京は日本の[MASK]です。
result_jp = unmasker("東京は日本の[MASK]です。")
print("=== 일본어 ===")
print("입력: 東京は日本の[MASK]です。")
for r in result_jp[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()
```

# 6 한국어 특화 BERT

```python
# 6번 빈칸 채우기 (한국어 특화 BERT)
from transformers import pipeline

unmasker = pipeline("fill-mask", model="klue/bert-base")

# 한국어 1: 대한민국의 수도는 [MASK]이다.
result1 = unmasker("대한민국의 수도는 [MASK]이다.")
print("=== 한국어 1 ===")
print("입력: 대한민국의 수도는 [MASK]이다.")
for r in result1[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 한국어 2: 삼성전자는 반도체와 [MASK]를 생산하는 글로벌 기업이다.
result2 = unmasker("삼성전자는 반도체와 [MASK]를 생산하는 글로벌 기업이다.")
print("=== 한국어 2 ===")
print("입력: 삼성전자는 반도체와 [MASK]를 생산하는 글로벌 기업이다.")
for r in result2[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 한국어 3: 인공지능 기술은 앞으로 [MASK] 분야에서 가장 큰 변화를 가져올 것이다.
result3 = unmasker("인공지능 기술은 앞으로 [MASK] 분야에서 가장 큰 변화를 가져올 것이다.")
print("=== 한국어 3 ===")
print("입력: 인공지능 기술은 앞으로 [MASK] 분야에서 가장 큰 변화를 가져올 것이다.")
for r in result3[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()
```

# 7 한국어 금융 특화 BERT

```python
# 7번 빈칸 채우기 (한국어 금융 특화 BERT)
from transformers import pipeline

unmasker = pipeline("fill-mask", model="snunlp/KR-FinBert")

# 금융 1: 한국은행이 기준금리를 [MASK] 결정했다.
result1 = unmasker("한국은행이 기준금리를 [MASK] 결정했다.")
print("=== 금융 1 ===")
print("입력: 한국은행이 기준금리를 [MASK] 결정했다.")
for r in result1[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 금융 2: 코스피 지수가 장중 [MASK]를 기록하며 투자자들의 관심이 집중되었다.
result2 = unmasker("코스피 지수가 장중 [MASK]를 기록하며 투자자들의 관심이 집중되었다.")
print("=== 금융 2 ===")
print("입력: 코스피 지수가 장중 [MASK]를 기록하며 투자자들의 관심이 집중되었다.")
for r in result2[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()

# 금융 3: 삼성전자 주가가 실적 발표 이후 [MASK] 전환했다.
result3 = unmasker("삼성전자 주가가 실적 발표 이후 [MASK] 전환했다.")
print("=== 금융 3 ===")
print("입력: 삼성전자 주가가 실적 발표 이후 [MASK] 전환했다.")
for r in result3[:3]:
    print(f"  예측: {r['token_str']} (score: {r['score']:.4f})")
print()
```
