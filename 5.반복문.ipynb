```python
# 문자열 순회
isUpper = False
isDigit = False
isSpecial = False
isLower = False
special_char = '!@#%?$'

password = 'Abc21!' 
#비밀번호 조건 만족여부 확인
for char in password :
    if char.isupper() : #대문자인지 확인
        isUpper = True
    elif char.isdigit() : #숫자인지 확인
        isDigit = True
    elif char.islower():
        isLower = True
    elif char in special_char :
        isSpecial = True
    check_result = isUpper + isDigit + isLower + isSpecial
print(f"비밀번호 강도 조건 만족여부 : { '만족' if check_result == 4 and isDigit else '불만족'}")
```

# 요일 구하기

```python
#날짜 입력받기 > 요일을 출력하는 기능
year = int(input('년도를 입력하시오: '))
month = int(input('월을 입력하시오: '))
day = int(input('일을 입력하시오: '))
print('수요일입니다')
```

```python
import datetime
calc_day = datetime.date(year,month,day).weekday() #요일값 0~ 월~

week_day = ['월','화','수','목','금','토','일']
#week_day[calc_day]
print(f'{week_day[calc_day]}요일입니다')
```

```python
#총일수 % 7 -> calc_day 
#입력일 : 2025-10-5
# 년도 : 1~2024 년도수 * 365(평년) | 366 (윤년)
# 월 : 1~9 월 * 30(4,6,9,11) | 31(1,3,5,7,8,10,12) | 28|29(2-윤년,평년) 
# 일 : 일
```

```python
year,month,day
```

```python
# 연도 날수계산
total_days = 0

for i in range(1,year): #1년부터 2024년까지
    if i % 400 == 0: #윤년
        total_days += 366
    elif i % 100 == 0 : #평년
        total_days += 365
    elif i % 4 == 0 : #윤년
        total_days += 366
    else : #평년
        total_days += 365 

print(f'1st : {total_days}, {year}')

# 2025년의 월기준 일수 계산
month_days = [0,31,28,31,30,31,30,31,31,30,31,30,31]
for m in range(1,month):
    total_days += month_days[m]
    
print(f'2nd : {total_days}, {year}')

#올해가 윤년인지 체크 +1
if month >= 3 :
    #윤년체크
    if year % 400 == 0: #윤년
        total_days += 1
    elif year % 100 == 0 : #평년
        total_days += 0
    elif year % 4 == 0 : #윤년
        total_days += 1
    else : #평년
        total_days += 0 

print(f'3nd : {total_days}, {year}, {month}')       

total_days += day

print(f'final : {total_days}, {year}, {month}, {day}')   

# 연도 + 월 + 일 % 7 > 요일에 해당하는 수
remain = total_days % 7
result = None

if remain == 0:
    result = '일요일'
elif remain == 1:
    result = '월요일'
elif remain == 2:
    result = '화요일'
elif remain == 3:
    result = '수요일'
elif remain == 4:
    result = '목요일'
elif remain == 5:
    result = '금요일'
elif remain == 6:
    result = '토요일'

print(f'{year}년 {month}월 {day}일은 {result}입니다.')   
```

# 성적리포트 출력

```python
students = [{'num':'1','name':'kim', 'kor':80,'math':90,'eng':100,},
            {'num':'2','name':'park', 'kor':90,'math':60,'eng':80,},
            {'num':'3','name':'hong', 'kor':70,'math':70,'eng':80,},
            {'num':'4','name':'hong1', 'kor':70,'math':70,'eng':80,},
            {'num':'5','name':'hong2', 'kor':70,'math':70,'eng':80,},
           ]
```

```python
#학생별 총점, 평균, 순위 -> 속성추가 -> 출력
#전체 총점, 평균 출력
#과목별 평균 출력
```

```python
# 반 총점, 평균 및 각 과목별 총점과 평균 초기화
class_total = 0
class_avg   = 0.0
kor_total   = 0
kor_avg     = 0.0
eng_total   = 0
eng_avg     = 0.0
math_total  = 0
math_avg    = 0.0


# 학생들의 성적 총점과 평균 및 반 총점과 과목별 총점을 구한다.
for student in students:
    student['total'] = student['kor'] + student['eng'] + student['math']
    student['avg']   = student['total'] / 3

    class_total = class_total + student['total']
    kor_total   = kor_total   + student['kor']
    eng_total   = eng_total   + student['eng']
    math_total  = math_total  + student['math']

# 학생들의 총점에 따른 순위 계산 (선택)
students_sorted = sorted(students, key=lambda x: x['total'], reverse=True)

for idx, student in enumerate(students_sorted, start=1):
    student['order'] = idx

class_avg = class_total / len(students)
kor_avg   = kor_total   / len(students)
eng_avg   = eng_total   / len(students)
math_avg  = math_total  / len(students)


# 학생별 성적 처리 결과를 출력한다.
for student in students:
    print('번호: {:s}, 이름: {:s}, 국어: {:d}, 영어: {:d}, 수학: {:d}, 총점: {:d}, 평균: {:.2f}'.format(
           student['num'], student['name'],
           student['kor'], student['eng'], student['math'],
           student['total'], student['avg'])
         )
print()


# 반 평균 및 각 과목별 평균을 출력한다.
print('반 평균  : {:.2f}'.format(class_avg))
print('국어 평균: {:.2f}'.format(kor_avg)  )
print('영어 평균: {:.2f}'.format(eng_avg)  )
print('수학 평균: {:.2f}'.format(math_avg) )
```

# 업앤다운 게임

```python
import random
number = random.randint(1,10)
number #컴퓨터의 선택
```

```python
# 정답 맞추기
cnt = 0
while True:
    #입력받기
    answer = int(input('1~10 사이의 숫자를 맞추세요'))
    cnt += 1
    #조건
    if answer == number :  #맞춘 경우
        print(f'축하합니다. 정답입니다. 도전횟수 : {cnt}')
        break
    elif answer > number :
        print('down')
    else:
        print('up')
        
```
