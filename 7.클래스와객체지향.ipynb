```python
# 자동차 클래스 정의
- self.속성명 : 인스턴스 속성
- (그냥)속성명 : 클래스 속성

- __inint__(self) 생성자(에서), 매개변수(로) = 인스턴스 속성 설정
- def 메소드(self): 메서드 정의

- 클래스 메소드(cls) : 클래 속성 관리, @classmethod "데코레이터"
- 정적 메소드 : 유틸리티 함수, @staticmethod "데코레이터"
```

```python
class car:
    #1 - 클래스 속성 (공통 속성)
    instance_cnt = 0  

    # def __init__(self) : # 생성자 정의
    #   print('자동차 객체가 생성되었습니다.')
    def __init__(self, color = '') : # 생성자 정의
        #2 - 객체의 속성
        self.color = color

        car.instance_cnt += 1

        print(f'{self.color} 자동차 객체가 생성되었습니다.({car.instance_cnt}대)')

    def auto_pilot(self):
        print(f'{self.color} 자동차가 자율주행중입니다.') 
        #00대의 자동차가 운행중입니다.
        print(f'{car.instance_cnt}대의 자동차가 운행중입니다.')
    @classmethod
    def count_check(cls):
        print(f'자동차 대수 : {cls.instance_cnt}대')

    @staticmethod
    def type_check(type_code): #유틸리티
      if type_code > 20 : 
          print('전기자동차')
      elif type_code < 20 :
          print('내연자동차')
      else :
          print('하이브리드')    
```

```python
#cal1 = Calculator([1,2,3,4,5])
class calculator:
    def __init__ (self, number_list):
        self.number_list = number_list


    #인스턴스 메소드
    def sum(self):
        total = sum(self.number_list)
        print(f'합계: {total}')
        return total


    #인스턴스 메소드
    def avg(self):
        avg_value = self.sum() / len(self.number_list)  
        print(f'평균: {avg_value:.2f}')
        
```

```python
c1 = calculator([1,2,3,4,5])
c1.number_list
```

```python
c2 = calculator([6,7,8,9,10])
c2.number_list
```

```python
Q2. Account 클래스 은행에 가서 계좌를 개설하면 은행이름, 예금주 계좌번호, 잔액이 설정됩니다.
Account 클래스를 생성한 후 생성자를 구현해보세요. 생성자에서는 예금주와 초기 잔액만 입력 받습니다.
은행 이름은 SC은행으로 계좌벊호는 3자리-2자리-6자리 형태로 랜덤하게 생성됩니다.

은행이름: SC은행
계좌번호: 111-11-111111
```

```python
import random
num1 = str(random.randint(0, 999)).zfill(3)
num2 = str(random.randint(0, 99)).zfill(2)
num3 = str(random.randint(0, 999999)).zfill(6)
f'{num1}-{num2}-{num3}'
```

```python
import random

class Account:
    def __init__(self, name, balance):
        self.name = name
        self.bank_name = 'SC은행'
        self.balance = balance
        # 호출하는 이름을 정의된 메소드 이름과 일치시킵니다.
        self.account_number = self.make_account_number()
      
    @staticmethod
    def make_account_number():
        # zfill을 사용하여 자리수를 맞춘 랜덤 계좌번호 생성
        num1 = str(random.randint(0, 999)).zfill(3)
        num2 = str(random.randint(0, 99)).zfill(2)
        num3 = str(random.randint(0, 999999)).zfill(6)
        return f'{num1}-{num2}-{num3}'  
    
    def displayAccount(self):
        print(f'은행명: {self.bank_name}')
        print(f'예금주: {self.name}')
        print(f'계좌번호: {self.account_number}')
        print(f'잔액: {self.balance:,}원')

# 인스턴스 생성 및 확인
acc1 = Account('james', 1000000)
acc1.displayAccount()
```

```python
# 인스턴스 메소드
def displayAccount(self):
    print(f'은행명: {self.bank_name}')
    print(f'계좌번호: {self.account_number}')
    print(f'잔액: {self.balance}')


# 입금 기능
def deposit(self, amount):
    self.balance += amount
    print(f'{amount}원이 입금되었습니다. 잔액은 {self.balance}원 입니다.')  

# 출금 기능
def withdraw(self, amount):
    if amount > self.balance:
        print('잔액이 부족합니다.')
    else:
        self.balance -= amount
        print(f'{amount}원이 출금되었습니다. 잔액은 {self.balance}원 입니다.')
```
