```python
# 1. 피벗 테이블 생성 (오류 수정: 인자 이름을 명확히 지정)
# values에는 수치 데이터(tip), index와 columns에는 범주형 데이터를 넣습니다.
pivot_df = df_tips.pivot_table(values="tip", index="sex", columns="time")

# 2. 히트맵 작성 (오류 수정: camp -> cmap)
sns.heatmap(pivot_df, annot=True, cmap='YlGnBu')
```

```python
# 흡연 여부(smoker)와 시간대(time)별 데이터 개수 히트맵
# 특정 조합에 얼마나 많은 데이터가 있는지 '빈도'를 보고 싶을 때 사용합니다.
```

```python
# pd.crosstab을 사용하면 범주형 변수 간의 빈도를 쉽게 구할 수 있습니다.
ct_df = pd.crosstab(df_tips['smoker'], df_tips['time'])

# 히트맵 출력
sns.heatmap(ct_df, annot=True, fmt='d', cmap='Oranges') # fmt='d'는 정수로 표시
```

```python
import folium

# 1. 서울 중심부의 위도와 경도로 지도 객체 생성
# location=[위도, 경도], zoom_start=확대 정도
seoul_map = folium.Map(location=[37.5665, 126.9780], zoom_start=12)

# 2. 서울의 주요 명소 5곳 데이터 (이름, 위도, 경도)
places = [
    {"name": "경복궁", "lat": 37.5796, "lon": 126.9770},
    {"name": "N서울타워", "lat": 37.5512, "lon": 126.9882},
    {"name": "롯데월드타워", "lat": 37.5125, "lon": 127.1025},
    {"name": "명동", "lat": 37.5594, "lon": 126.9860},
    {"name": "북촌한옥마을", "lat": 37.5829, "lon": 126.9835}
]

# 3. 반복문을 사용하여 지도에 마커 추가
for place in places:
    folium.Marker(
        location=[place["lat"], place["lon"]],
        popup=place["name"],  # 클릭 시 나타나는 문구
        tooltip=f"여기는 {place['name']}입니다!",  # 마우스를 올렸을 때 나타나는 문구
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(seoul_map)

# 4. 결과 지도를 HTML 파일로 저장하거나 주피터 노트북에서 바로 확인
seoul_map.save('seoul_landmarks.html')
seoul_map
```

```python
import folium
import requests

# 출발지: 노량진역
start = [126.9416, 37.5138]   # [경도, 위도]

# 도착지: 지정 위치
end = [127.502497, 37.666399]  # [경도, 위도]

# OSRM API 요청 (자동차 기준)
url = f"http://router.project-osrm.org/route/v1/driving/{start[0]},{start[1]};{end[0]},{end[1]}?overview=full&geometries=geojson"

response = requests.get(url)
data = response.json()

# 경로 좌표 추출 (위도, 경도 순서로 변환)
route_coords = [
    [lat, lon] for lon, lat in data["routes"][0]["geometry"]["coordinates"]
]

# 지도 생성 (중간 지점 기준)
m = folium.Map(
    location=[(37.5138 + 37.666399) / 2, (126.9416 + 127.502497) / 2],
    zoom_start=10
)

# 출발지 마커
folium.Marker(
    location=[37.5138, 126.9416],
    popup="노량진역",
    icon=folium.Icon(color="blue", icon="train")
).add_to(m)

# 도착지 마커
folium.Marker(
    location=[37.666399, 127.502497],
    popup="목적지",
    icon=folium.Icon(color="red", icon="flag")
).add_to(m)

# 최적 경로 라인
folium.PolyLine(
    locations=route_coords,
    color="green",
    weight=5,
    opacity=0.8,
    tooltip="최적 경로"
).add_to(m)

m
```
