```python
import tensorflow as tf
from sklearn import linear_model #머신러닝 사이킷런 라이브러리
import numpy 
```

```python
from scipy.linalg import null_space
#딥러닝
X_train = numpy.array([1,2,3,4,5,6,7,]).reshape(-1,1) #문제
y_train = numpy.array([12,17,22,27,32,37,42]).reshape(-1,1) #정답

#깡통준비 - 모델 구성
inputs = tf.keras.layers.Input(shape=(1,)) #입력층 정의
outputs = tf.keras.layers.Dense(units=1)(inputs) #출력층

model = tf. keras.models.Model(inputs, outputs)
model.compile(loss = tf.keras.losses.MeanSquaredError(),
              optimizer = tf.keras.optimizers.SGD()) #깡통준비
model.fit(X_train, y_train, epochs=1000, verbose=0) #학습
model.predict(numpy.array([[9]]))
```
